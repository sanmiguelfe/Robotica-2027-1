## Introducción

En el desarrollo de proyectos de robótica es necesario utilizar diferentes herramientas de software para **crear archivos, automatizar tareas, documentar procedimientos y compartir el trabajo**. Aunque estas herramientas tienen funciones diferentes, su combinación permite establecer flujos de trabajo reproducibles y facilita la colaboración.

- <img src="imagenes/Ubuntu-logo.png" height=60/> **Ubuntu** es un sistema operativo basado en Linux ampliamente utilizado en robótica. El manejo de la terminal permite ejecutar programas, administrar archivos y automatizar tareas sin depender exclusivamente de interfaces gráficas.

- **Bash (Bourne Again SHell)** es un intérprete de comandos utilizado habitualmente en sistemas Linux. Además de permitir ejecutar comandos de manera interactiva, permite crear **scripts de Bash**, es decir, archivos de texto que contienen una secuencia de comandos que puede ejecutarse de manera automatizada.

- <img src="imagenes/Jupyter-logo.png" height=60/> **Jupyter** proporciona un entorno interactivo basado en notebooks. Un notebook está formado por celdas que pueden contener texto, ecuaciones, código y resultados. En esta práctica se utilizará para documentar el trabajo y ejecutar comandos de Bash desde el mismo documento.

- **Markdown** es un lenguaje de marcado ligero que permite dar formato al texto mediante una sintaxis sencilla. Jupyter utiliza Markdown para crear títulos, listas, tablas, enlaces y ecuaciones dentro de las celdas de texto.

- <img src="imagenes/Git-logo.png" height=60/> **Git** es un sistema de control de versiones que permite registrar cambios en los archivos de un proyecto, recuperar versiones anteriores y trabajar mediante ramas.

- <img src="imagenes/Github-logo.png" height=60/> **GitHub** es una plataforma que permite alojar repositorios Git en línea y facilita el trabajo colaborativo mediante repositorios, ramas, colaboradores y herramientas de revisión.

En robótica, estas herramientas pueden utilizarse conjuntamente. Por ejemplo, un alumno puede crear un script de Bash para preparar una carpeta de trabajo, documentar el procedimiento en Jupyter y finalmente almacenar el notebook y los scripts en un repositorio de GitHub.


## Cuestionario previo

### Responder de forma breve las siguientes preguntas:

- En Linux, ¿qué es una **shell** y qué relación existe entre `sh` y `bash`?

> Una shell es un intérprete que permite comunicarse con el sistema operativo mediante comandos. `sh` es el nombre de la shell tradicional y de una interfaz compatible con el estándar POSIX, mientras que Bash es una shell que incorpora características adicionales. No siempre corresponden al mismo programa, por lo que un script que utiliza funciones específicas de Bash debe ejecutarse con Bash.

- ¿Qué es un **script de Bash** y cuál es la diferencia entre ejecutar un comando directamente en la terminal y ejecutarlo mediante un script?

> Un script de Bash es un archivo de texto que contiene instrucciones ejecutadas por el intérprete Bash. En la terminal, los comandos se introducen de manera interactiva; en un script, se guardan para ejecutar una secuencia de forma repetible. Esto permite automatizar tareas y utilizar condiciones y ciclos.

- ¿Qué es **Jupyter** y qué tipo de información puede contener una celda de un notebook?

> Jupyter es un entorno interactivo que permite crear documentos llamados notebooks. Sus celdas pueden contener texto con formato, ecuaciones, imágenes o código ejecutable. Las celdas de código también pueden mostrar resultados como texto, tablas y gráficas.

- ¿Qué es **Markdown** y para qué se utiliza dentro de Jupyter?

> Markdown es un lenguaje de marcado ligero que permite dar formato y estructura al texto mediante símbolos sencillos. Dentro de Jupyter se utiliza para escribir títulos, listas, tablas, enlaces e insertar imágenes. Además, las celdas Markdown permiten presentar ecuaciones mediante sintaxis LaTeX.

- ¿Cuál es la diferencia entre **Git** y **GitHub**?

> Git es un sistema de control de versiones que registra los cambios de los archivos de un proyecto y permite recuperar versiones anteriores. GitHub es una plataforma que aloja repositorios Git en línea y facilita la colaboración. Git puede utilizarse localmente sin GitHub; para enviar los commits locales al repositorio remoto se utiliza `git push`.

### 1. Markdown

Crear una celda de Markdown que contenga:

- La forma general de la matriz de transformación homogénea de un sistema $j$ respecto a un sistema $i$.
- La forma general de la matriz de rotación de un sistema $j$ respecto a un sistema $i$ respecto al eje $z$.
- La forma general del vector de posición de un sistema $j$ respecto a un sistema $i$.
- La forma general del vector de postura para un manipulador contenido en el plano $XY$.
- La forma general del vector de juntas para un manipulador RRR.

Utilizar la sintaxis de **LaTeX** disponible en Markdown para escribir las ecuaciones.


#### Matriz de transformación homogénea

La matriz de transformación homogénea describe la posición y orientación del sistema j respecto al sistema i. Permite transformar coordenadas expresadas en j a coordenadas expresadas en i:

$$
{}^{i}\mathbf{T}_{j} =
\begin{bmatrix}
{}^{i}\mathbf{R}_{j} & {}^{i}\mathbf{p}_{j} \\
\mathbf{0}_{1\times3} & 1
\end{bmatrix}
=
\begin{bmatrix}
r_{11} & r_{12} & r_{13} & p_x \\
r_{21} & r_{22} & r_{23} & p_y \\
r_{31} & r_{32} & r_{33} & p_z \\
0 & 0 & 0 & 1
\end{bmatrix}
$$

- ${}^{i}\mathbf{R}_{j}$ representa la orientación y tiene tamaño $3\times3$.
- ${}^{i}\mathbf{p}_{j}$ indica la posición del origen de j, expresada en el sistema i, y tiene tamaño $3\times1$.
- La transformación completa tiene tamaño $4\times4$.

#### Matriz de rotación respecto al eje z

Para una orientación relativa que corresponde únicamente a una rotación de ángulo $\theta$ alrededor del eje z:

$$
{}^{i}\mathbf{R}_{j}=\mathbf{R}_{z}(\theta)=
\begin{bmatrix}
\cos\theta & -\sin\theta & 0 \\
\sin\theta & \cos\theta & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

El ángulo se expresa en radianes. Su sentido positivo sigue la regla de la mano derecha y la componente z permanece sin cambio.

#### Vector de posición

El vector de posición contiene las coordenadas del origen de j respecto al sistema i, expresadas en los ejes de i:

$$
{}^{i}\mathbf{p}_{j}=
\begin{bmatrix}
{}^{i}x_{j} \\
{}^{i}y_{j} \\
{}^{i}z_{j}
\end{bmatrix}
$$

Las tres componentes representan posiciones sobre los ejes x, y y z. Las componentes se expresan en unidades de longitud, por ejemplo, metros.

#### Vector de postura de un manipulador en el plano XY

La postura del efector final P respecto al sistema de referencia 0 se representa mediante dos coordenadas de posición y un ángulo de orientación:

$$
{}^{0}\boldsymbol{\xi}_{P}(\mathbf{q})=
\begin{bmatrix}
{}^{0}x_{P}(\mathbf{q}) \\
{}^{0}y_{P}(\mathbf{q}) \\
{}^{0}\alpha_{P}(\mathbf{q})
\end{bmatrix}
$$

- ${}^{0}x_{P}$ y ${}^{0}y_{P}$ indican la posición del efector final.
- ${}^{0}\alpha_{P}$ representa su orientación alrededor del eje z.
- $\mathbf{q}$ contiene las variables articulares del manipulador.

La tercera componente es un ángulo, no una coordenada de altura.

#### Vector de juntas de un manipulador RRR

Un manipulador RRR tiene tres articulaciones de revolución. Su vector de variables articulares es:

$$
\mathbf{q}=
\begin{bmatrix}
\theta_{1} \\
\theta_{2} \\
\theta_{3}
\end{bmatrix}
$$

Cada componente representa el ángulo de una articulación respecto a su configuración de referencia y se expresa en radianes.

El vector de juntas describe la configuración de las articulaciones, mientras que el vector de postura describe la posición y orientación del efector final.


### 2. Manejo de Ubuntu y Bash

En esta primera parte se creará un **script de Bash** que automatice la creación y eliminación de una estructura de carpetas.

El script deberá:

- Desplegar un mensaje de inicio.
- Moverse a la carpeta de usuario mediante `~` o `$HOME`.
- Verificar si existe la carpeta `Practica1`.
- Si no existe, crearla.
- Dentro de `Practica1`, crear la carpeta `Letras` con los archivos:
  - `a.txt`
  - `b.txt`
  - `c.txt`
- Dentro de `Practica1`, crear la carpeta `Integrantes` con los archivos correspondientes a los integrantes del equipo.
- Mostrar un diagrama de la estructura de `Practica1` utilizando el comando `tree`.
- Eliminar la carpeta `Practica1`, incluyendo sus subcarpetas y archivos.
- Dar permisos de ejecución al script y ejecutarlo.

**Importante:** antes de utilizar comandos que eliminen archivos, se comprueba que la ruta corresponda únicamente a la carpeta creada para esta práctica.

#### Probatorios

En la siguiente captura se muestra la ejecución del script de Bash desarrollado para la práctica:

![Ejecución del script](imagenes/screenshot.png)

### Código del script

```bash
#!/bin/bash

# Mensaje de inicio
echo "Inicio de la Practica 1"

# Moverse a la carpeta del usuario
cd "$HOME"

# Verificar si existe la carpeta Practica1
if [ -d "$HOME/Practica1" ]; then
    echo "La carpeta Practica1 ya existe"
else
    echo "La carpeta Practica1 no existe. Se creara."
    mkdir "$HOME/Practica1"
fi

# Crear carpeta Letras y sus archivos
mkdir -p "$HOME/Practica1/Letras"
touch "$HOME/Practica1/Letras/a.txt"
touch "$HOME/Practica1/Letras/b.txt"
touch "$HOME/Practica1/Letras/c.txt"

# Crear carpeta Integrantes y sus archivos
mkdir -p "$HOME/Practica1/Integrantes"
touch "$HOME/Practica1/Integrantes/Daniel.txt"
touch "$HOME/Practica1/Integrantes/Fernando.txt"
touch "$HOME/Practica1/Integrantes/Monse.txt"
touch "$HOME/Practica1/Integrantes/Othon.txt"

# Mostrar la estructura creada
echo ""
echo "Estructura de Practica1:"
tree "$HOME/Practica1"

# Verificar la carpeta antes de eliminarla
echo ""
echo "Verificando carpeta antes de eliminar..."

if [ -d "$HOME/Practica1" ]; then
    echo "La carpeta a eliminar es: $HOME/Practica1"
    rm -r "$HOME/Practica1"
    echo "Practica1 fue eliminada correctamente"
fi

echo "Fin de la Practica 1"
```

### Descripción de los bloques principales

- **Inicio:** muestra un mensaje indicando el comienzo de la práctica y se mueve al directorio principal del usuario mediante `$HOME`.

- **Verificación de `Practica1`:** comprueba si la carpeta existe. Si no existe, se crea utilizando `mkdir`.

- **Carpeta `Letras`:** crea la carpeta `Letras` y los archivos `a.txt`, `b.txt` y `c.txt` utilizando los comandos `mkdir` y `touch`.

- **Carpeta `Integrantes`:** crea la carpeta `Integrantes` y genera un archivo `.txt` para cada integrante del equipo: Daniel, Fernando, Monse y Othon.

- **Visualización:** utiliza el comando `tree` para mostrar de forma organizada la estructura de carpetas y archivos que fue generada por el script.

- **Verificación antes de eliminar:** mediante la condición `if [ -d "$HOME/Practica1" ]` se comprueba nuevamente que la carpeta `Practica1` exista antes de ejecutar el comando de eliminación.

- **Eliminación:** después de verificar la existencia y ruta de la carpeta, se utiliza `rm -r` para eliminar `Practica1` junto con todos sus archivos y subcarpetas.

- **Finalización:** finalmente se muestra el mensaje `Fin de la Practica 1`, indicando que todas las instrucciones del script terminaron de ejecutarse.

## Análisis de resultados

Responder las siguientes preguntas con base en el trabajo realizado:

1. ¿Qué ventajas ofrece un script de Bash frente a ejecutar manualmente una secuencia de comandos en la terminal?

> Un script de Bash permite guardar una serie de comandos dentro de un archivo para poder ejecutarlos de manera automática y en el mismo orden cada vez que sea necesario. Durante la práctica se observó que esto facilita tareas repetitivas, como crear carpetas y archivos, comprobar si un directorio existe o eliminar una estructura de directorios. Si todos estos comandos se escribieran manualmente en la terminal habría que introducirlos uno por uno y existiría una mayor posibilidad de cometer errores. Además, el script puede modificarse, reutilizarse y compartirse con otros usuarios, lo que permite automatizar procedimientos que normalmente tomarían más tiempo.

3. ¿Qué diferencias existen entre una celda de Markdown y una celda de código en Jupyter?

> Una celda de Markdown se utiliza principalmente para agregar texto y organizar la información dentro de un notebook. Permite utilizar títulos, listas, tablas, imágenes, enlaces y expresiones matemáticas mediante LaTeX. Esto fue útil en la práctica para escribir y presentar las matrices y vectores utilizados en robótica. En cambio, una celda de código contiene instrucciones que pueden ser ejecutadas por el kernel de Jupyter y puede generar una salida como resultado. Por lo tanto, las celdas Markdown sirven principalmente para documentar y explicar el trabajo, mientras que las celdas de código permiten realizar y ejecutar operaciones dentro del mismo notebook.

4. ¿Qué utilidad tiene un repositorio en línea para almacenar scripts, notebooks y documentación de un proyecto de robótica?

> Un repositorio en línea permite mantener organizados en un mismo lugar los diferentes archivos que forman parte de un proyecto, como scripts, notebooks, imágenes y documentación. Al utilizar Git y GitHub también es posible llevar un registro de los cambios realizados mediante commits, lo que facilita conocer cómo ha ido evolucionando el proyecto y recuperar versiones anteriores en caso de ser necesario. Otra ventaja importante es el trabajo colaborativo, ya que diferentes integrantes pueden tener acceso al repositorio y contribuir al mismo proyecto. En robótica esto resulta útil debido a que normalmente se trabaja con diferentes tipos de archivos y con proyectos que se van modificando y ampliando conforme avanza su desarrollo.
